```python
import re
import os

def update_file(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        content = f.read()
    
    # Regex to find the intro object at the start of the array
    # It looks like: { type: 'intro', ... },
    # We capture the content inside the braces
    intro_pattern = r"\{\s*type:\s*'intro',(.*?)\},"
    
    match = re.search(intro_pattern, content, re.DOTALL)
    if not match:
        print(f"No intro found in {filepath}")
        return

    intro_body = match.group(1)
    full_intro_match = match.group(0)
    
    # Extract fields
    def get_field(name):
        m = re.search(f"{name}:\s*(['\"].*?['\"]|\[.*?\])", intro_body, re.DOTALL)
        return m.group(1) if m else None

    title = get_field('title')
    intro_content = get_field('content')
    duration = get_field('duration')
    level = get_field('level')
    objectives = get_field('objectives')
    
    # Construct introConfig object string
    intro_config = "introConfig: {\n"
    if title: intro_config += f"    title: {title},\n"
    if intro_content: intro_config += f"    content: {intro_content},\n"
    if duration: intro_config += f"    duration: {duration},\n"
    if level: intro_config += f"    level: {level},\n"
    if objectives: intro_config += f"    objectives: {objectives},\n"
    intro_config += "  },"

    # Find the next object start
    # The intro object is followed by comma, then newline/space, then {
    rest_of_content = content[match.end():]
    next_obj_start = rest_of_content.find('{')
    
    if next_obj_start == -1:
        print(f"No next object found in {filepath}")
        return

    # Insert introConfig into the next object
    # We insert it right after the opening brace
    insertion_point = match.end() + next_obj_start + 1
    
    new_content = content[:match.start()] + content[match.end():match.end()+next_obj_start+1] + "\n  " + intro_config + content[insertion_point:]
    
    # Clean up potential double commas or spacing issues if needed
    # The regex removal leaves the trailing comma of the intro object?
    # My regex included the trailing comma `},`
    # So `content[:match.start()]` ends before the intro.
    # `content[match.end():]` starts after the intro comma.
    # So we effectively removed the intro object.
    
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(new_content)
    print(f"Updated {filepath}")

# Test on l1-1.js
# update_file(r"c:\Users\Chungshu\Desktop\bordeaux-map\src\data\lessons\l1-1.js")
```
